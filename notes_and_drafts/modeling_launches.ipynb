We can use the poisson distribution to model the number of launches per month.
If the mean nuber of launches per month is a function of production rates say 1 ship per day per line and we have 3 lines then the mean number of launches per month is 90.
Let's say we want to know the probability of having 90 launches in a month.

```python
import math

λ = 90
k = 90

probability = (λ**k) * math.exp(-λ) / math.factorial(k)
print(probability)

# or like this

from scipy.stats import poisson

probability = poisson.pmf(90, mu=90)
print(probability)
```

We can visualize the probability of different values for k by looping through the values of k and calculating the probability.
or we can use the probability mass function of the poisson distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

λ = 90
k_values = np.arange(0, 200)
pmf_values = poisson.pmf(k_values, mu=λ)

import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(go.Bar(
    x=k_values,
    y=pmf_values,
    marker_color='skyblue'
))

fig.update_layout(
    title=f"Poisson Distribution (λ = {λ})",
    xaxis_title="Number of Starship Launches",
    yaxis_title="Probability",
    xaxis=dict(range=[60, 120]),  # Zoom in around the interesting region
    template="plotly_white"
)

fig.show()
```

We can also compute the cumulative distribution function of the poisson distribution. This allows us to answer questions like what the probability of having 90 or fewer launches in a month.

```python
poisson.cdf(90, mu=90)
```

```python
1 - poisson.cdf(99, mu=90) # Chance of 100+ launches in a month
```
